```python
import pandas as pd
import dash
from dash import dcc, html
import plotly.express as px

# Load your CSV data
df = pd.read_csv('Data/household_power_consumption.txt',  sep=';', low_memory=False)
# Combine Date and Time into a single DateTime column
df['DateTime'] = pd.to_datetime(df['Date'] + ' ' + df['Time'], format='%d/%m/%Y %H:%M:%S')

# Convert columns to numeric
for col in ['Sub_metering_1','Sub_metering_2','Sub_metering_3','Global_active_power']:
    df[col] = pd.to_numeric(df[col], errors='coerce').fillna(0)

# Initialize the Dash app
app = dash.Dash(__name__)

# Create figures
fig_active_power = px.line(df, x='DateTime', y='Global_active_power',
                           title='Global Active Power Over Time')

fig_sub_metering = px.line(df, x='DateTime', y=['Sub_metering_1','Sub_metering_2','Sub_metering_3'],
                           title='Energy Sub-Metering Comparison')

# Layout
app.layout = html.Div(children=[
    html.H1(children='Hydro One Energy Dashboard'),

    html.Div(children='''Interactive dashboard for energy consumption data.'''),

    dcc.Graph(id='active-power-graph', figure=fig_active_power),
    dcc.Graph(id='sub-metering-graph', figure=fig_sub_metering)
])

if __name__ == '__main__':
    app.run(debug=True)
```
